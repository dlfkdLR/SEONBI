## 질문 수별 MAE 비교 그림 (선택)

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(9, 3.2), sharey=True)
for ax, name in zip(axes, ["purchase", "career"], strict=True):
    r = sim.run_experiment(name, USERS, SEED)
    for s in sim.STRATEGIES:
        ax.errorbar(
            sim.FIXED_COUNTS,
            [r["fixed"][k][s]["mae"]["mean"] for k in sim.FIXED_COUNTS],
            yerr=[r["fixed"][k][s]["mae"]["se"] for k in sim.FIXED_COUNTS],
            marker="o",
            label=s,
        )
    ax.set_title(name)
    ax.set_xlabel("questions")
    ax.legend()
axes[0].set_ylabel("MAE")
plt.tight_layout()
```
